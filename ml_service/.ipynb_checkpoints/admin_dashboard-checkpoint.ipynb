# Admin Dashboard - Machine Learning
Utilisez ce notebook pour réentraîner le modèle et visualiser les performances.

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pymongo import MongoClient
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, classification_report
import joblib
from dotenv import load_dotenv

# Charger les variables d'environnement
load_dotenv(os.path.join(os.getcwd(), '..', '.env'))
```

## 1. Connexion à la Base de Données / Chargement des Données
Nous allons récupérer les talents depuis MongoDB ou générer des données factices si la base est vide.

```python
MONGO_URI = os.getenv('MONGO_URI', 'mongodb://localhost:27017/talentcard')

def generate_mock_data():
    data = [
        {"skills": ["React", "Node.js", "MongoDB"], "projects": ["E-commerce site", "Chat app"], "languages": ["English", "French"], "verified": True},
        {"skills": ["HTML", "CSS"], "projects": ["Portfolio"], "languages": ["English"], "verified": False},
        {"skills": ["Python", "Data Science", "Pandas"], "projects": ["Stock predictor"], "languages": ["English", "German"], "verified": True},
        {"skills": [], "projects": [], "languages": [], "verified": False},
        {"skills": ["Java", "Spring"], "projects": ["Bank system"], "languages": ["French"], "verified": True},
        {"skills": ["C++"], "projects": ["Game"], "languages": ["English"], "verified": True},
        {"skills": ["Word", "Excel"], "projects": [], "languages": ["French"], "verified": False},
        {"skills": ["Javascript", "React"], "projects": ["Todo list"], "languages": ["English"], "verified": True},
        {"skills": ["Unknown"], "projects": [], "languages": [], "verified": False},
        {"skills": ["Go", "Docker", "Kubernetes"], "projects": ["Microservices"], "languages": ["English", "Spanish"], "verified": True},
    ]
    return pd.DataFrame(data)

try:
    client = MongoClient(MONGO_URI)
    db = client.get_database()
    talent_collection = db.talents
    talents = list(talent_collection.find())
    if len(talents) < 5:
        print("Pas assez de données en DB, utilisation de données factices...")
        df = generate_mock_data()
    else:
        df = pd.DataFrame(talents)
    print("Données chargées :", len(df), "entrées")
except Exception as e:
    print(f"Erreur de connexion MongoDB ({e}), utilisation de données factices...")
    df = generate_mock_data()
```

## 2. Préparation des Données et Entraînement
Nous combinons les compétences, projets et langues en une seule caractéristique textuelle pour le modèle.

```python
def preprocess(df):
    df['text_features'] = df.apply(lambda x: ' '.join(x.get('skills', [])) + ' ' +
                                             ' '.join(x.get('projects', [])) + ' ' +
                                             ' '.join(x.get('languages', [])), axis=1)
    return df

df = preprocess(df)
X = df['text_features']
y = df['verified']

if len(df) < 5:
    X_train, X_test, y_train, y_test = X, X, y, y
else:
    test_size = 0.3 if len(df) > 20 else 0.5
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=test_size, random_state=42)

pipeline = Pipeline([
    ('tfidf', TfidfVectorizer(stop_words='english', max_features=1000)),
    ('clf', RandomForestClassifier(n_estimators=100, random_state=42))
])

pipeline.fit(X_train, y_train)
print("Modèle entraîné avec succès.")
```

## 3. Évaluation et Visualisation
Affichage de la matrice de confusion pour évaluer la performance du modèle.

```python
y_pred = pipeline.predict(X_test)
cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['Prédit: Non', 'Prédit: Oui'],
            yticklabels=['Réel: Non', 'Réel: Oui'])
plt.title(f'Matrice de Confusion (Précision: {pipeline.score(X_test, y_test):.2f})')
plt.ylabel('Vrai Label')
plt.xlabel('Label Prédit')
plt.show()
```

## 4. Sauvegarde du Modèle
Enregistrez le modèle entraîné pour qu'il soit utilisé par l'application.

```python
model_path = 'model.pkl'
joblib.dump(pipeline, model_path)
print(f"Modèle sauvegardé dans {os.path.abspath(model_path)}")
```
